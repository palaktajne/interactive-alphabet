```python
# ============================================================
# AUTOMATED CONTENT FILTER & PLAGIARISM DETECTOR
# MODULE 1: TEXT INPUT & PREPROCESSING
# ============================================================

# Install required libraries
!pip install -q gradio python-docx nltk

# Import libraries
import re
import gradio as gr
import nltk
from docx import Document

# Download NLTK resource
nltk.download('punkt', quiet=True)


# ============================================================
# FUNCTION 1: CLEAN TEXT
# ============================================================

def preprocess_text(text):

    # Convert text to lowercase
    text = text.lower()

    # Remove extra spaces, tabs and new lines
    text = re.sub(r'\s+', ' ', text)

    # Remove spaces from beginning and end
    text = text.strip()

    return text


# ============================================================
# FUNCTION 2: COUNT WORDS
# ============================================================

def get_word_count(text):

    if not text:
        return 0

    words = text.split()

    return len(words)


# ============================================================
# FUNCTION 3: COUNT CHARACTERS
# ============================================================

def get_character_count(text):

    return len(text)


# ============================================================
# FUNCTION 4: COUNT SENTENCES
# ============================================================

def get_sentence_count(text):

    if not text:
        return 0

    sentences = re.split(r'[.!?]+', text)

    # Remove empty sentences
    sentences = [s.strip() for s in sentences if s.strip()]

    return len(sentences)


# ============================================================
# FUNCTION 5: ANALYZE TEXT
# ============================================================

def analyze_text(text):

    # Check whether user entered text
    if not text or not text.strip():

        return {
            "cleaned_text": "",
            "word_count": 0,
            "character_count": 0,
            "sentence_count": 0
        }

    # Clean the text
    cleaned_text = preprocess_text(text)

    # Calculate statistics
    word_count = get_word_count(cleaned_text)
    character_count = get_character_count(cleaned_text)
    sentence_count = get_sentence_count(cleaned_text)

    return {
        "cleaned_text": cleaned_text,
        "word_count": word_count,
        "character_count": character_count,
        "sentence_count": sentence_count
    }


# ============================================================
# FUNCTION 6: PROCESS USER INPUT
# ============================================================

def process_text(text):

    # Check empty input
    if not text or not text.strip():

        return (
            "Please enter or paste some text.",
            "",
            0,
            0,
            0
        )

    # Analyze text
    result = analyze_text(text)

    return (
        "Text successfully processed ✓",
        result["cleaned_text"],
        result["word_count"],
        result["character_count"],
        result["sentence_count"]
    )


# ============================================================
# GRADIO WEBSITE
# ============================================================

with gr.Blocks(
    title="Automated Content Filter & Plagiarism Detector"
) as app:

    # Project heading
    gr.Markdown("""
    # 🔍 Automated Content Filter & Plagiarism Detector

    ## Module 1 — Text Input & Preprocessing

    Enter or paste your document text below and click
    **Analyze Text** to process the document.
    """)

    # Text input
    text_input = gr.Textbox(
        label="📝 Enter or Paste Your Text",
        placeholder="Type or paste your document text here...",
        lines=12
    )

    # Analyze button
    analyze_button = gr.Button(
        "🔎 Analyze Text",
        variant="primary"
    )

    # Status
    status = gr.Textbox(
        label="📌 Status"
    )

    # Cleaned text
    cleaned_output = gr.Textbox(
        label="🧹 Cleaned Text",
        lines=8
    )

    # Statistics
    gr.Markdown("### 📊 Text Statistics")

    with gr.Row():

        word_output = gr.Number(
            label="📝 Word Count"
        )

        character_output = gr.Number(
            label="🔤 Character Count"
        )

        sentence_output = gr.Number(
            label="📄 Sentence Count"
        )

    # Connect button with function
    analyze_button.click(
        fn=process_text,
        inputs=text_input,
        outputs=[
            status,
            cleaned_output,
            word_output,
            character_output,
            sentence_output
        ]
    )


# ============================================================
# LAUNCH APPLICATION
# ============================================================

app.launch(share=True)
```
